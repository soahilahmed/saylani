```python
def detecting_outlaer(col):
    q1=np.percentile(col,25)
    q3=np.percentile(col,75)
    IQR=q3-q1
    lower_boundry=q1-1.5*IQR
    uper_boundry=q3+1.5*IQR
    return lower_boundry,uper_boundry
```

```python
lb,ub=detecting_outlaer(df.price)
```

```python
df[(df.price<lb)|(df.price>ub)]
```
