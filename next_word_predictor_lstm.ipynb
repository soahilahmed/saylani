```python
faqs = """
About Houses and Properties

What is the average price of a 3-bedroom house?
The average price depends on the city and location. In major cities, it can range from $150,000 to $500,000.

What documents are required to buy a house?
You generally need a valid ID, proof of income, bank statements, sale agreement, and property ownership documents.

Can foreigners buy property?
Yes, in many countries foreigners can buy property, but regulations vary depending on local laws.

What is a down payment?
A down payment is the initial amount paid upfront when purchasing a property. It is usually between 10% and 20% of the property's price.

What is a mortgage?
A mortgage is a loan provided by a bank or financial institution to help you purchase a property.

How long does it take to buy a house?
The process usually takes between 30 and 90 days depending on financing and legal procedures.

What is property tax?
Property tax is an annual tax imposed by the government based on the value of your property.

What is freehold ownership?
Freehold ownership means you own both the property and the land permanently.

What is leasehold ownership?
Leasehold ownership means you own the property for a fixed number of years but not the land.

Can I negotiate the property price?
Yes, in most cases buyers can negotiate with the seller before signing the agreement.

Rental Properties

What is the average rent for a 2-bedroom apartment?
The rent depends on the location, facilities, and demand in the area.

Is a security deposit required?
Yes, landlords usually require one or two months' rent as a refundable security deposit.

Can pets be allowed in rental properties?
It depends on the landlord's policy.

Who is responsible for maintenance?
Minor maintenance is usually the tenant's responsibility, while major repairs are handled by the landlord.

Can I terminate the lease early?
Yes, but there may be penalties depending on the lease agreement.

Property Investment

Is real estate a good investment?
Real estate is generally considered a long-term investment that can generate rental income and property appreciation.

What factors increase property value?
Location, nearby schools, transportation, infrastructure, property condition, and market demand.

What is ROI in real estate?
ROI (Return on Investment) measures the profit earned from a property compared to its total investment cost.

Should I invest in residential or commercial property?
Residential properties are generally lower risk, while commercial properties may offer higher returns but require larger investments.

Home Loans

What is the minimum credit score for a home loan?
Most banks prefer a credit score above 650, although requirements vary.

Can I pay off my mortgage early?
Yes, but some lenders may charge a prepayment fee.

How is the interest rate calculated?
Interest rates depend on market conditions, loan amount, loan tenure, and your credit profile.

Property Inspection

Why is a property inspection important?
It helps identify structural issues, plumbing problems, electrical faults, and other defects before purchase.

What should I check before buying a house?
Check the foundation, roof, plumbing, electrical system, legal documents, neighborhood, and market value.

Selling Property

How can I sell my property faster?
Price it competitively, use high-quality photos, market it online, and ensure all legal documents are ready.

Do I need a real estate agent?
No, but an experienced agent can help with pricing, marketing, negotiations, and paperwork.

Legal Questions

How do I verify property ownership?
Verify the title deed and check government land records before purchasing.

Can a property have multiple owners?
Yes, a property can be jointly owned by two or more individuals.

What is a title deed?
A title deed is the legal document proving ownership of a property.

What happens after the sale agreement?
The buyer pays the remaining amount, ownership is transferred, and the property is officially registered.

Maintenance

How often should a house be maintained?
Basic maintenance should be done every year, while major inspections are recommended every 5 to 10 years.

What is included in home insurance?
Home insurance typically covers fire, theft, natural disasters, and accidental damage depending on the policy.

Can solar panels increase property value?
Yes, energy-efficient upgrades like solar panels can increase a property's market value.

"""
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer=Tokenizer()
tokenizer.fit_on_texts([faqs])
```

```python
##totall length of vocubalar or sentence
```

```python
len(tokenizer.word_index)
```

```python
# text to sequence
```

```python
input_sequences=[]
for sentence in faqs.split('\n'):
    sentence = sentence.strip() # ya line senternce ka between ma space ko remove kar rahi

    if sentence:
      tokenized_sentence=tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1,len(sentence)):
      input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
len_max=max([len(x) for x in input_sequences])
```

```python
len_max
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padding_input=pad_sequences(input_sequences,maxlen=len_max,padding='pre')
```

```python
X=padding_input[:,:-1]
```

```python
y=padding_input[:,-1]
```

```python
X.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=len(tokenizer.word_index)+1)
```

```python
y.shape
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,LSTM,Embedding
```

```python
# NEURA NETWOR LSTM
```

```python
model=Sequential()
model.add(Embedding(input_dim=len(tokenizer.word_index)+1,output_dim=100,input_length=len_max-1))
model.add(LSTM(150, return_sequences=True))
model.add(LSTM(150))
model.add(Dense(len(tokenizer.word_index)+1,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=10)
```

```python
import numpy as np
import time
text = "What is a down payment"

for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=len_max, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(2)
```
